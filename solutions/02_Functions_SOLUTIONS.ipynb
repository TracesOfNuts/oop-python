## 1. What is a function?

Think of a **vending machine**.

- You put something **in** (money and a button number).
- The machine does some work inside. You do not see how.
- Something comes **out** (a drink).

A function is the same:

- You give it some values (the **inputs**).
- It runs some lines of code.
- It can give you back a result (the **output**).

You already know some functions that Python gives you: `print()`, `len()`, `int()`, `type()`.
Now you will learn to write your own.

## 2. Defining and calling a function

Here is a very small function:

```python
def showMenu():
    print("Milk Tea   $4.80")
    print("Green Tea  $3.90")
```

Let us read it piece by piece:

- `def` means "I am **def**ining a new function".
- `showMenu` is the name we chose. Use camelCase, like variables.
- `()` the brackets hold the inputs. This function needs none, so they are empty.
- `:` the colon says "the body starts on the next line".
- The body is the lines **indented** under it (4 spaces). They belong to the function.

⚠️ **Defining a function does not run it.**
It only teaches Python a new word.
To run it, you **call** it by writing its name with brackets: `showMenu()`.

````python
def showMenu():
    print("Milk Tea   $4.80")
    print("Green Tea  $3.90")

print("Before the call")
showMenu()
print("After the call")
````

You can call a function as many times as you like.
That is the point of a function: write it once, use it many times.

````python
showMenu()
showMenu()
````

## 3. Parameters: giving the function some input

Most functions need some information to do their job.
You put a name inside the brackets of the `def` line.
This name is a **parameter**: a variable that gets its value when the function is called.

````python
def greet(name):
    print(f"Hello, {name}!")

greet("Amy")
greet("Ben")
````

When we call `greet("Amy")`:

1. Python jumps into the function.
2. The parameter `name` gets the value `"Amy"`.
3. The body runs, and prints `Hello, Amy!`.
4. Python comes back and carries on after the call.

Two words that are easy to mix up:

- **Parameter**: the name in the `def` line (`name`).
- **Argument**: the actual value you pass in when you call it (`"Amy"`).

## 4. Several parameters

A function can have several parameters.
Separate them with commas.
When you call it, the arguments are matched **in the same order**.

````python
def showOrder(drink, cups):
    print(f"{cups} x {drink}")

showOrder("Milk Tea", 2)     # drink gets "Milk Tea", cups gets 2
showOrder("Coffee", 1)
````

## 5. `return`: giving a value back

So far our functions only **print**.
Printing shows something on the screen, but the rest of your program cannot use it.

Remember the vending machine.
Printing is like the machine showing a picture of a drink on its screen.
`return` is the machine **dropping the real drink** into your hand, so you can use it.

````python
def double(number):
    return number * 2

result = double(4)
print(result)
print(double(10) + 1)
````

When Python reaches `return`:

1. It works out the value after `return`.
2. It leaves the function **immediately**.
3. The function call is replaced by that value.

So `double(4)` becomes `8`, and `result = double(4)` is the same as `result = 8`.

🎯 Most exam methods **return** a value, for example "returns the total fees" or "returns True if added".
If the question says "returns", you must use `return`, not `print`.

## 6. Using returned values

Because `return` gives back a real value, you can use the call anywhere you could use a value:
in maths, in an f-string, or as the argument of another function.

````python
def priceOf(cups):
    return cups * 4.80

total = priceOf(2) + priceOf(3)
print(f"Total: ${total:.2f}")
print(f"One cup: ${priceOf(1):.2f}")
````

## 8. Default values for parameters

You can give a parameter a **default value** with `=` in the `def` line.
If the caller leaves that argument out, the default is used.

````python
def topUp(balance, amount=10):
    return balance + amount

print(topUp(5))        # amount is not given, so it is 10
print(topUp(5, 50))    # amount is given, so it is 50
````

⚠️ Parameters with defaults must come **after** the ones without defaults.
`def topUp(amount=10, balance):` is a `SyntaxError`.

🎯 **Why this matters for the exam.**
Some languages let you write two functions with the same name but different parameters.
This is called **overloading**.
Python does not allow that, so we use default values instead.
The course slides do this with `def deduct(self, amount=5):`.
The exam has asked you to fix "overloading" code using default values.

## 10. Common mistakes and their error messages

Without `()`, the function is not called.
Python just shows that the name is a function.

````python
def getMessage():
    return "Hello"

print(getMessage)      # no brackets: prints something like <function getMessage at 0x...>
print(getMessage())    # with brackets: calls it and prints Hello
````

Read the message: it tells you which argument is missing.

````python
def greet(name):
    print(f"Hello, {name}!")

greet()
````

`TypeError: greet() missing 1 required positional argument: 'name'`

**Forgetting the colon** at the end of the `def` line.

````python
def greet(name)
    print(f"Hello, {name}!")
````

**Forgetting to indent** the body.

````python
def greet(name):
print(f"Hello, {name}!")
````

Python reads from top to bottom, so the `def` must run first.

## 📝 Key points

- `def name(parameters):` defines a function. The body is indented.
- Defining does not run it. Call it with brackets: `name(arguments)`.
- Arguments are matched to parameters by position, or by name (`amount=50`).
- `return` gives a value back and leaves the function at once.
- A function with no `return` gives back `None`.
- Default values: `def deduct(balance, amount=5):`. This is how Python does "overloading".
- Variables made inside a function do not exist outside it.

| Mistake | What happens |
| --- | --- |

| `getMessage` instead of `getMessage()` | The function is not called |
| Wrong number of arguments | `TypeError: ... missing 1 required positional argument` |
| No `:` at the end of the `def` line | `SyntaxError` |
| Body not indented | `IndentationError` |
| Using a local variable outside the function | `NameError` |
| Calling before the `def` has run | `NameError` |
