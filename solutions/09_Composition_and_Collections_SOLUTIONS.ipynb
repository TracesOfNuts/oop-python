## 1. "Has a" or "is a"?

There are two ways to reuse a class.

**Composition** means "**has a**".
One object keeps another object inside it.

- A laptop **has a** hard disk.
- A customer **has a** cash card.
- A phone book **has** contacts.

**Inheritance** means "**is a**".
One class is a more specific kind of another class.
You will learn it in the next notebook.

- A manager **is an** employee.
- A member cash card **is a** cash card.

💡 Say the sentence out loud.
If "A ___ has a ___" sounds right, it is composition.
If "A ___ is a ___" sounds right, it is inheritance.

## 2. An object inside an object

Here is a `CashCard` class like the one from the course.
Run it, because the next examples use it.

````python
class CashCard:
    def __init__(self, id, amount):
        self.__id = id
        self.__balance = amount

    @property
    def id(self):
        return self.__id

    @property
    def balance(self):
        return self.__balance

    def topUp(self, amount):
        if amount > 0:
            self.__balance += amount

    def deduct(self, amount):
        if self.__balance >= amount:
            self.__balance -= amount
            return True
        return False

    def __str__(self):
        return f"Id: {self.__id} Balance: ${self.__balance:.2f}"
````

A customer **has a** cash card.
The simplest way: create the card first, then pass it in to the customer.
The customer stores the **card object** in an attribute.

````python
class Customer:
    def __init__(self, name, card):
        self.__name = name
        self.__card = card          # the Customer HAS A CashCard

    @property
    def name(self):
        return self.__name

    @property
    def card(self):
        return self.__card

c = CashCard("1", 10.0)        # 1. create the card first
cust = Customer("James", c)    # 2. give it to the customer
print(cust.name)
print(cust.card)               # prints the card, using CashCard's __str__
print(cust.card.balance)       # two levels: customer -> card -> balance
````

Read `cust.card.balance` from left to right:

1. `cust.card` gives the `CashCard` object inside the customer.
2. `.balance` gives that card's balance.

```text
   cust (Customer)                    c (CashCard)
   __name = "James"                   __id      = "1"
   __card = ----------------------->  __balance = 10.0
```

The customer does not hold a copy of the card.
It holds a **link** to the same card object.

## 3. Delegation: passing the work on

Usually the outside world talks to the customer, not to the card.
So the customer gets its own methods that **pass the work on** to the card.
This is called **delegation**.

```python
def topUp(self, amount):
    self.__card.topUp(amount)      # ask the card to do it

def cardBalance(self):
    return self.__card.balance     # ask the card for its balance
```

There is also a second way to build the customer: create the card **inside** `__init__`.
Then the code outside never has to make a `CashCard` itself.

```python
class Customer:
    def __init__(self, name, cardId, amount):
        self.__name = name
        self.__card = CashCard(cardId, amount)   # created inside

cust = Customer("Jack", "2", 50)
```

Which way to use depends on the question.
Read the class diagram or the sample code to see what the constructor takes.

Here is a `Harddisk` class.
Run it for the next exercise.

````python
class Harddisk:
    def __init__(self, capacity):
        self.__capacity = capacity
        self.__used = 0

    @property
    def free(self):
        return self.__capacity - self.__used

    def save(self, size):
        if size <= self.free:
            self.__used += size
            return True
        return False
````

## 4. A class that holds a list of objects

This is the most important pattern in this notebook.
A `PhoneBook` keeps many `Contact` objects in a list.

Run the `Contact` class first.

````python
class Contact:
    def __init__(self, name, phone):
        self.__name = name
        self.__phone = phone

    @property
    def name(self):
        return self.__name

    @property
    def phone(self):
        return self.__phone

    @phone.setter
    def phone(self, phone):
        self.__phone = phone

    def __str__(self):
        return f"Name: {self.__name} Phone: {self.__phone}"
````

The phone book creates an **empty list** in `__init__`.
Every new phone book gets its own new list.

````python
class PhoneBook:
    def __init__(self, owner):
        self.__owner = owner
        self.__contacts = []          # a NEW empty list for every phone book

    def addContact(self, contact):
        self.__contacts.append(contact)

    def count(self):
        return len(self.__contacts)

book = PhoneBook("Jim")
book.addContact(Contact("Peter", "91231234"))
book.addContact(Contact("Amy", "61232310"))
print(book.count())
````

⚠️ **Always create the list inside `__init__`.**
If you put the list somewhere else, all the objects can end up **sharing one list**.
These two predictions show how.

## 5. Search, remove and update

These three methods appear again and again in the exam.
Learn them as patterns.

**Search** returns the object, or `None` if it is not there:

```python
def searchContact(self, name):
    for c in self.__contacts:
        if c.name == name:
            return c        # found it: stop and give it back
    return None             # AFTER the loop: we checked everyone
```

⚠️ The `return None` must be **after** the loop, not in an `else` inside the loop.
Otherwise the method gives up after looking at the first contact only.

**Remove** and **update** first search, then act, and return `True` or `False`.

### 🏋️ Exercise 9.12: Search a contact

````python
class PhoneBook:
    def __init__(self, owner):
        self.__owner = owner
        self.__contacts = []

    def addContact(self, contact):
        self.__contacts.append(contact)

    def searchContact(self, name):
        for c in self.__contacts:
            if c.name == name:
                return c
        return None
````

### 🏋️ Exercise 9.13: Remove a contact

Add `removeContact(name)`.

- If there is no contact with that name, return `False`.
- Otherwise remove that contact from the list and return `True`.

Reuse `searchContact`, so you do not have to write the loop again.

````python
class PhoneBook:
    def __init__(self, owner):
        self.__owner = owner
        self.__contacts = []

    def addContact(self, contact):
        self.__contacts.append(contact)

    def searchContact(self, name):
        for c in self.__contacts:
            if c.name == name:
                return c
        return None

    def removeContact(self, name):
        c = self.searchContact(name)
        if c is None:
            return False
        self.__contacts.remove(c)
        return True
````

### 🏋️ Exercise 9.25: Total price, and no repeated pizzas

Improve `PizzaOrder`:

- `addPizza(pizza, qty)`: if a pizza with the **same name** is already in the order, add `qty` to that line instead of adding a new line.
- `totalPrice()` returns the sum of `price * qty` for every line.
- `lineCount()` returns how many lines (different pizzas) there are.

````python
class PizzaOrder:
    def __init__(self, customer):
        self.__customer = customer
        self.__lines = []

    def addPizza(self, pizza, qty):
        for line in self.__lines:
            if line[0].name == pizza.name:
                line[1] += qty
                return
        self.__lines.append([pizza, qty])

    def totalQty(self):
        total = 0
        for line in self.__lines:
            total += line[1]
        return total

    def totalPrice(self):
        total = 0
        for line in self.__lines:
            total += line[0].price * line[1]
        return total

    def lineCount(self):
        return len(self.__lines)
````

## 📝 Key points

- **Composition** is "has a": one object stores another object in an attribute.
- **Inheritance** is "is a": one class is a special kind of another class.
- Delegation: the outer object's method calls the inner object's method, for example `self.__card.topUp(amount)`.
- Create the list (or dictionary) inside `__init__`, so every object gets its own.
- Search: loop, `return` the object when found, and `return None` **after** the loop.
- Remove and update: search first, then return `True` or `False`.
- Maximum size: `if len(self.__items) + 1 > ClassName._MAX: ...`
- Case insensitive: compare `a.lower() == b.lower()`.
- Totals: start at 0, loop, add the field of each object.
- A dictionary of objects uses a unique key, such as an email, and `.get(key)` to find.

## ⚠️ Common mistakes

| Mistake | What happens |
| --- | --- |
| `return None` inside the loop, in an `else` | Only the first item is ever checked |
| A list as a class variable or a default parameter | All objects share one list |
| `len(self.__items) > _MAX` before adding | Lets one extra item in |
| Forgetting `return` in `pay` or `removeActivity` | The caller gets `None` |
| `self.__card.balance()` with brackets | `TypeError`, because `balance` is a property |
