## 1. Why classes?

Think of a **form** for a cash card application.
The empty form has boxes: "Card id" and "Balance".
That empty form is like a **class**.
It describes what every cash card has, but it is not a real card yet.

When you fill in the form, you get a real card, for example id `123` with balance `$20`.
A filled in form is like an **object**.
You can fill in the same form many times, so one class can make many objects.

| Word | Meaning | Example |
| --- | --- | --- |
| **class** | the blueprint or template | `CashCard` |
| **object** (or **instance**) | one real thing made from the class | the card with id `123` |
| **attribute** (or **instance variable**) | data that each object stores | `id`, `balance` |
| **method** | something the object can do | `topUp`, `deduct` |

Putting the data and the actions together in one class is called **abstraction**.
We keep only what matters for our program.
A cash card in our program needs an id and a balance, but not its colour.

## 2. Your first class

Here is a class for a cash card.

````python
class CashCard:
    def __init__(self, id, amount):
        self.id = id
        self.balance = amount

c1 = CashCard("123", 20.0)
c2 = CashCard("456", 10.0)
print(c1.id, c1.balance)
print(c2.id, c2.balance)
````

Let us read it slowly.

- `class CashCard:` starts the class. Class names use CapitalWords.
- `def __init__(self, id, amount):` is the **constructor**.
  It runs automatically every time you create a new object.
  Its job is to fill in the attributes.
  (`__init__` has **two** underscores on each side.)
- `self.id = id` means "store the value of `id` in **this object**, under the name `id`".
- `c1 = CashCard("123", 20.0)` creates a new object and stores it in `c1`.

**The arguments go in the same order as the constructor parameters, after `self`.**
`"123"` goes into `id`, and `20.0` goes into `amount`.
You never pass `self` yourself. Python does it for you.

## 3. What is `self`?

`self` means **"the object we are working on right now"**.

When you write `c1 = CashCard("123", 20.0)`, this is what happens:

1. Python makes a new, empty object.
2. Python calls `__init__` and puts that new object into `self`.
   So `self` **is** the new object, and `id` is `"123"`, and `amount` is `20.0`.
3. `self.id = id` writes `"123"` into the new object.
4. `self.balance = amount` writes `20.0` into the new object.
5. The finished object is stored in `c1`.

Then `CashCard("456", 10.0)` does the same again, but now `self` is a **different** new object.
You can picture it like this:

```text
c1  ---->  [ id: "123"   balance: 20.0 ]
c2  ---->  [ id: "456"   balance: 10.0 ]
```

Each object has its own boxes.
`self.balance` means "the balance box of the object we are working on".

💡 The word `self` is just a name, but everyone uses `self`, and so must you in the exam.

## 4. Methods: what an object can do

A **method** is a function written inside a class.
Its first parameter is always `self`, so the method knows which object it is working on.

````python
class CashCard:
    def __init__(self, id, amount):
        self.id = id
        self.balance = amount

    def topUp(self, amount):
        self.balance += amount

    def deduct(self, amount):
        if self.balance >= amount:
            self.balance -= amount

c1 = CashCard("123", 20.0)
c1.topUp(10)
print(c1.balance)
c1.deduct(5)
print(c1.balance)
````

When you write `c1.topUp(10)`, Python puts `c1` into `self` and `10` into `amount`.
So inside the method, `self.balance += amount` means "add 10 to **c1's** balance".

You call a method with **dot** and **brackets**: `object.method(arguments)`.
The course calls this "sending a message to an object".

## 6. Making `print(obj)` useful with `__str__`

What happens if you print an object?

````python
class CashCard:
    def __init__(self, id, amount):
        self.id = id
        self.balance = amount

c1 = CashCard("123", 20.0)
print(c1)
````

You see something like `<__main__.CashCard object at 0x000001F2A3B4C5D0>`.
That is Python saying "this is a CashCard object, stored at this memory address".
The number changes every time, and it is not useful to a person.

To fix this, write a special method called `__str__`.
It must **return** a string that describes the object.
Then `print(obj)` uses that string automatically.

````python
class CashCard:
    def __init__(self, id, amount):
        self.id = id
        self.balance = amount

    def __str__(self):
        return f"Id: {self.id} Balance: ${self.balance:.2f}"

c1 = CashCard("123", 20.0)
print(c1)
print(str(c1))
message = f"My card -> {c1}"
print(message)
````

🎯 **In the exam**, almost every class needs a `__str__`.
The paper shows the exact output, for example `Id: 123 Balance: $20.00`.
Copy the spaces, colons and `$` exactly. Formatting is part of the marks.

⚠️ `__str__` must **return** the string, not print it.
If it prints and returns nothing, you get an error:

````python
class Broken:
    def __str__(self):
        print("I am broken")

print(Broken())
````

Read the last line: `TypeError: __str__ returned non-string (type NoneType)`.
The method printed, but returned `None`, and `print()` needs a string.

## 10. Common mistakes (and their error messages)

**Mistake 1: forgetting `self` in the method definition.**

````python
class Light:
    def __init__(self):
        self.isOn = False

    def switchOn():            # no self!
        self.isOn = True

lamp = Light()
lamp.switchOn()
````

`TypeError: Light.switchOn() takes 0 positional arguments but 1 was given`.
Python always passes the object into the first parameter.
`switchOn` has no parameter to receive it, so Python complains.
**Fix:** `def switchOn(self):`

**Mistake 2: forgetting `self.` in front of an attribute.**

````python
class Light:
    def __init__(self):
        self.isOn = False

    def isItOn(self):
        return isOn            # should be self.isOn

lamp = Light()
print(lamp.isItOn())
````

`NameError: name 'isOn' is not defined`.
Inside a method, `isOn` alone is a normal variable, and there is none.
The attribute lives in the object, so you must write `self.isOn`.

**Mistake 3: the wrong number of arguments when creating an object.**

````python
class CashCard:
    def __init__(self, id, amount):
        self.id = id
        self.balance = amount

c = CashCard("123")
````

`TypeError: CashCard.__init__() missing 1 required positional argument: 'amount'`.
The constructor wants `id` **and** `amount`.

**Mistake 4: forgetting the brackets when calling a method.**

````python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

r = Rectangle(4, 3)
print(r.area)      # no brackets: the method is NOT called
print(r.area())    # with brackets: the method runs
````

Without brackets you get something like `<bound method Rectangle.area of ...>`.
The method was not run. Always use `()` to call a method.

## 📝 Key points

- A **class** is a template. An **object** is one thing made from it.
- `__init__(self, ...)` is the constructor. It runs when you create an object and sets the attributes.
- `self` is the object being worked on. Write `self.name` to use an attribute inside the class.
- Create an object with `ClassName(arguments)`. The arguments follow the constructor order, without `self`.
- Every method has `self` as its first parameter.
- Call a method with `obj.method(arguments)`, always with brackets.
- `__str__` must **return** a string. Then `print(obj)` shows it.
- A loop over a list of objects gives you one object at a time.

## ⚠️ Common mistakes

| Mistake | What happens |
| --- | --- |
| `def topUp(amount):` (no `self`) | `TypeError: ... takes 1 positional argument but 2 were given` |
| `balance += 5` instead of `self.balance += 5` | `NameError` or `UnboundLocalError` |
| `CashCard("123")` when the constructor needs 2 values | `TypeError: ... missing 1 required positional argument` |
| `print(r.area)` | Prints `<bound method ...>` instead of calling it |
| `__str__` that prints instead of returns | `TypeError: __str__ returned non-string` |
| `_init_` with one underscore | The constructor never runs |
