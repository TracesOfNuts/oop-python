## 1. What is a dictionary?

Think of a real phone book.
You look up a **name** (the key) and you get a **phone number** (the value).
You never ask "what is entry number 3?". You ask "what is Joe's number?".

A dictionary works the same way.
You write it with curly brackets `{ }`.
Each pair is written `key: value`, and pairs are separated by commas.

````python
phones = {"Peter": "91231234", "Joe": "81231234", "Amy": "61232310"}
print(phones)
print(len(phones))      # number of pairs
````

More examples:

```python
fees = {"Business": 500, "Computing": 650, "Arts": 400}   # major -> fees
stock = {"apple": 12, "pear": 0}                            # fruit -> how many
empty = {}                                                  # an empty dictionary
```

- **Keys** are usually strings (sometimes numbers). Each key appears only **once**.
- **Values** can be anything: numbers, text, lists, even objects.

## 2. Reading a value

Put the **key** in square brackets.

````python
fees = {"Business": 500, "Computing": 650, "Arts": 400}
print(fees["Computing"])
````

⚠️ If the key does not exist, Python stops with a **KeyError**.

````python
fees = {"Business": 500, "Computing": 650, "Arts": 400}
print(fees["Medicine"])
````

Read the last line: `KeyError: 'Medicine'`.
It means "there is no key called Medicine".

💡 Keys are case sensitive. `fees["arts"]` is also a `KeyError`, because the key is `"Arts"`.

**The safe way: `.get()`**

`.get(key)` gives the value, or `None` if the key is missing.
`.get(key, default)` gives `default` instead of `None` when the key is missing.

````python
fees = {"Business": 500, "Computing": 650, "Arts": 400}
print(fees.get("Arts"))
print(fees.get("Medicine"))
print(fees.get("Medicine", 0))
````

🎯 Template from past papers:

```python

    return cls._FEES.get(key)   # None if missing
```

## 3. Adding and changing values

The same line does both jobs:

```python
fees["Law"] = 700
```

- If the key `"Law"` is **new**, the pair is **added**.
- If the key `"Law"` **already exists**, its value is **replaced**.

````python
fees = {"Business": 500, "Computing": 650}
fees["Law"] = 700          # new key: added
print(fees)
fees["Business"] = 550     # existing key: updated
print(fees)
````

You can also change a number value with `+=`, as long as the key exists.

````python
stock = {"apple": 12, "pear": 0}
stock["apple"] -= 2
stock["pear"] += 10
print(stock)
````

🎯 Past paper template:


## 4. Checking and removing keys

- `key in d` gives `True` if the key exists. It checks **keys**, not values.
- `del d[key]` removes the pair. It raises a `KeyError` if the key is missing.
- `d.pop(key)` removes the pair and gives back the value.

````python
phones = {"Peter": "91231234", "Joe": "81231234"}
print("Joe" in phones)
print("81231234" in phones)    # False: in checks keys, not values
if "Joe" in phones:
    del phones["Joe"]
print(phones)
````

💡 Same idea as lists: **check with `in` first**, then remove.

```python
if uid in cards:
    del cards[uid]
```

## 5. Looping over a dictionary

There are three ways, depending on what you need.

````python
fees = {"Business": 500, "Computing": 650, "Arts": 400}

for major in fees:                  # 1. the keys
    print(major)

for amount in fees.values():        # 2. only the values
    print(amount)

for major, amount in fees.items():  # 3. key AND value together
    print(f"{major}: ${amount:.2f}")
````

- `for key in d:` gives the **keys**.
- `for value in d.values():` gives the **values**.
- `for key, value in d.items():` gives **both**. You name two loop variables, separated by a comma.

💡 `.items()` is the most useful one. When in doubt, use it.

🎯 The course slides loop over a dictionary of objects with `for c in cards.values():`.

## 6. Recipes with dictionaries

The list recipes from notebook 04 work here too.

**Total of the values**

````python
budget = {"Food": 300, "Transport": 120, "Fun": 80}
total = 0
for amount in budget.values():
    total += amount
print(total)
````

**Counting with a dictionary**

To count how many times each thing appears, use the thing as the key and the count as the value.

````python
votes = ["Tea", "Coffee", "Tea", "Juice", "Tea"]
counts = {}
for drink in votes:
    if drink in counts:
        counts[drink] += 1
    else:
        counts[drink] = 1
print(counts)
````

💡 Why the `if`? The first time we see `"Tea"`, there is no key yet, so `counts["Tea"] += 1` would be a `KeyError`.
So the first time we set it to `1`.

### 🏋️ Exercise 5.15: Count the letters

`grades` is a list like `["A", "B", "A", "C", "A"]`.
Return a dictionary with how many times each grade appears, for example `{"A": 3, "B": 1, "C": 1}`.

````python
def countGrades(grades):
    counts = {}
    for grade in grades:
        if grade in counts:
            counts[grade] += 1
        else:
            counts[grade] = 1
    return counts

print(countGrades(["A", "B", "A", "C", "A"]))
````

### 🏋️ Exercise 5.16: Find the most expensive item

Complete `mostExpensive(prices)`.
`prices` is a dictionary of item to price.
Return the **key** (the item name) with the highest price.
Assume the dictionary is not empty.

````python
def mostExpensive(prices):
    bestItem = None
    bestPrice = 0
    for item, price in prices.items():
        if bestItem is None or price > bestPrice:
            bestItem = item
            bestPrice = price
    return bestItem

print(mostExpensive({"Tea": 2.5, "Cake": 6.0, "Bun": 1.8}))
````

3. `planReport(plan)` returns one string with one line per category, then the total, like this:

```text

```

No new line at the very end.

````python
def setBudget(plan, category, amount):
    plan[category] = amount

def spend(plan, category, amount):
    if category not in plan:
        return False
    if plan[category] < amount:
        return False
    plan[category] -= amount
    return True

def planReport(plan):
    lines = []
    total = 0
    for category, amount in plan.items():
        lines.append(f"{category}: ${amount:.2f}")
        total += amount
    lines.append(f"Total: ${total:.2f}")
    return "\n".join(lines)

myPlan = {}
setBudget(myPlan, "Food", 300)
setBudget(myPlan, "Fun", 100)
spend(myPlan, "Fun", 20)
print(planReport(myPlan))
````

## 📝 Key points

- A dictionary maps **keys** to **values**: `fees = {"Arts": 400}`.
- Read: `fees["Arts"]` (`KeyError` if missing) or `fees.get("Arts")` (`None` if missing).
- `fees.get("Law", 0)` gives a default when the key is missing.
- `fees["Law"] = 700` **adds** a new key or **updates** an existing one.
- `"Law" in fees` checks the **keys**. `del fees["Law"]` removes a pair.
- Loop: `for key in d`, `for value in d.values()`, `for key, value in d.items()`.
- A tuple `(date, status)` is a fixed pair. Unpack it with `date, status = pair`.

## ⚠️ Common mistakes

| Mistake | What happens |
| --- | --- |
| `fees["Medicine"]` when the key is missing | `KeyError` |
| `fees["arts"]` when the key is `"Arts"` | `KeyError` (keys are case sensitive) |
| `counts[x] += 1` the first time you see `x` | `KeyError` (set it to 1 first) |
| `20 in ages` to look for a value | `False` (`in` checks keys only) |
| `for k, v in d:` without `.items()` | `ValueError` (too many values to unpack) |
| Changing a tuple: `pair[1] = "x"` | `TypeError` |
