class Ticket(ABC):

### ✍️ Exercise 15.9: Q2(d) Types of overriding (5 marks)

(i) What type of overriding is `PremiumTicket.__str__()`? Explain. (2 marks)

(ii) What type of overriding is `PremiumTicket.price()`? Explain. (2 marks)

(iii) In the sample code, `print(t)` gives a different result for different tickets. What is this OOP idea called? (1 mark)

✍️ **My answer:** *(double-click here and type your answer)*

<details>
<summary>👀 Model answer (click to open)</summary>

(i) **Overriding by refinement.**
`PremiumTicket.__str__()` has the same name and parameters as `Ticket.__str__()`, and it calls the parent version with `super().__str__()`, then adds more text.

(ii) **Overriding by replacement.**
`PremiumTicket.price()` overrides `Ticket.price()` but does **not** call `super().price()`.
It provides its own complete calculation (here it implements the abstract method).

(iii) **Polymorphism.**
The same call `print(t)` (which calls `t.__str__()` and `t.price()`) runs a different method depending on the class of the object `t`.

            self.__show(f"Error: {e}")
