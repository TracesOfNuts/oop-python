## 2. Variables

A **variable** is a name that holds a value.
Think of it as a labelled box.
You put a value in the box with `=`.

```python
price = 4.50
```

Read this as "**price becomes 4.50**", not "price equals 4.50".
The `=` sign means "put the value on the right into the name on the left".

````python
drink = "Milk Tea"
price = 4.50
print(drink)
print(price)
````

A variable can be changed later.
The new value **replaces** the old one.

````python
cups = 2
print(cups)
cups = 5
print(cups)
````

The right side is worked out **first**, then stored in the name on the left.
So this line is fine, and very common:

````python
score = 10
score = score + 5     # take the old score (10), add 5, store 15 back in score
print(score)
````

💡 `score = score + 5` can also be written `score += 5`.
You will see `+=` and `-=` a lot in the exam, for example `self.__balance -= amount`.

**Naming rules**

- Names can use letters, digits and `_`, but cannot start with a digit.
- No spaces. Use **camelCase** for names with several words: `totalPrice`, `numberOfCups`.
- Names are case sensitive: `price` and `Price` are different variables.

## 3. Kinds of values (data types)

Every value has a **type**.
These are the ones you need:

| Type | Name in Python | Examples |
| --- | --- | --- |
| Whole number | `int` | `5`, `-3`, `0` |

`type(x)` tells you the type of `x`.

⚠️ `"5"` (with quotes) is **text**, not a number.
`True` and `False` start with a capital letter and have no quotes.

`None` means "no value".
You will use it later, for example a search that finds nothing returns `None`.

## 4. Doing maths

| Symbol | Meaning | Example | Result |
| --- | --- | --- | --- |
| `+` | add | `7 + 2` | `9` |
| `-` | subtract | `7 - 2` | `5` |
| `*` | multiply | `7 * 2` | `14` |
| `/` | divide (always gives a float) | `7 / 2` | `3.5` |
| `//` | divide and drop the decimal part | `7 // 2` | `3` |
| `%` | remainder | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |

Python follows normal maths order: brackets first, then `**`, then `* / // %`, then `+ -`.

````python
print(2 + 3 * 4)       # 3 * 4 first, so 14
print((2 + 3) * 4)     # brackets first, so 20
print(10 / 2)          # / always gives a float: 5.0
````

## 5. Working with text (strings)

You can join strings with `+`.
This is called **concatenation**.

````python
firstName = "Siti"
lastName = "Rahman"
fullName = firstName + " " + lastName
print(fullName)
print(len(fullName))    # len() counts the characters, including the space
````

⚠️ **You cannot add text and a number with `+`.**
Python does not know if you want maths or joining.
Run this to see the error.

````python
age = 19
print("Age: " + age)
````

Read the last line: `TypeError: can only concatenate str (not "int") to str`.
It means: "I can only join text to text, but you gave me an int."

The fix is to turn the number into text with `str()`:

````python
age = 19
print("Age: " + str(age))
````

## 6. f-strings: the easy way to build messages

Joining with `+` and `str()` gets messy.
An **f-string** is much easier.
Put an `f` before the quotes, and put variables inside curly brackets `{ }`.

````python
name = "Ravi"
cups = 2
message = f"{name} bought {cups} cups"
print(message)
````

**Money with 2 decimal places.**
Add `:.2f` inside the brackets.
It means "show this as a decimal number with 2 digits after the point".

````python
price = 4.5
print(f"Price: ${price:.2f}")
total = 14.4
print(f"Total: ${total:.2f}")
````

🎯 **In the exam**, `__str__` methods almost always need this, for example:

```text
Id: 123 Balance: $20.00
```

The `$` is just a normal character in the text.
The `{balance:.2f}` part is the number.

You can also put maths inside the brackets:

````python
price = 4.8
cups = 3
print(f"{cups} cups cost ${price * cups:.2f}")
````

💡 You will also see the older `.format()` style in exam code.
It does the same job.
The `{}` are filled in order with the values inside `.format(...)`.

````python
name = "Ravi"
price = 4.5
print("Name: {} Price: ${:.2f}".format(name, price))
````

## 7. Converting between types

| Function | Turns a value into | Example | Result |
| --- | --- | --- | --- |
| `int(x)` | a whole number | `int("42")` | `42` |
| `float(x)` | a decimal number | `float("4.5")` | `4.5` |
| `str(x)` | text | `str(42)` | `"42"` |

⚠️ If the text is not a number, `int()` and `float()` fail with a **ValueError**.

````python
print(int("abc"))
````

**`input()` always gives you text.**
`input()` asks the user to type something.
Even if they type a number, you get a string, so you must convert it.

````python
cups = input("How many cups? ")
print(type(cups))
cups = int(cups)
print(cups * 4.80)
````

🎯 In the GUI part of the exam, you read text from an entry box with `.get()`.
It is also always text, so you must use `float()` or `int()` before doing maths.
This is one of the most common exam mistakes.

## 📝 Key points

- `=` puts a value into a variable. Read it as "becomes".
- `+=` and `-=` update a variable: `balance -= 5`.
- Types: `int` (5), `float` (4.5), `str` ("text"), `bool` (True/False), `None`.
- `/` always gives a float. `//` drops the decimals. `%` gives the remainder.
- You cannot `+` a string and a number. Use `str()` or an f-string.
- f-strings: `f"Total: ${total:.2f}"` shows money with 2 decimal places.
- `input()` and GUI `.get()` always give text. Convert with `int()` or `float()`.

## ⚠️ Common mistakes

| Mistake | What happens |
| --- | --- |
| `print("Age: " + 19)` | `TypeError` |
| `int("4.5")` | `ValueError` (use `float()` for decimals) |
| Forgetting the `f` in `f"{x}"` | Prints `{x}` literally |
| `{price:.2f}` without the `$` | Prints `4.50` instead of `$4.50` |
| `true` instead of `True` | `NameError` |
