## How to answer "error or output?"

For every line, ask yourself these questions in order.

1. **Which object is this?** What class was it created from?
2. **Can the object be created?** An abstract class, or a subclass that still has an abstract method, cannot be created.
3. **Where does Python find the method or attribute?** Look in the object's own class first, then go up to the parent, then the grandparent.
4. **Does the name exist?** A private `__name` cannot be read from outside. A missing `super().__init__()` means the parent attributes were never made.
5. **Is it a property or a method?** A property is used without `()`. A method needs `()`.

If there is an error, the exam wants three things:

- **Which line** causes it.
- **The type of error**, for example `TypeError` or `AttributeError`.
- **Why**, in one sentence.

If something was printed before the error, write that output too.

## 1. Abstract classes

Run this cell first.
It defines the classes used by the next four exercises.
Read them carefully, just like in the exam.

````python
from abc import ABC, abstractmethod

class Appliance(ABC):
    def __init__(self, brand):
        self._brand = brand

    @property
    def brand(self):
        return self._brand

    @abstractmethod
    def power(self):
        pass

    @abstractmethod
    def describe(self):
        pass

class Fan(Appliance):
    def power(self):
        return 50

    def describe(self):
        return f"{self._brand} fan uses {self.power()}W"

class Heater(Appliance):
    def power(self):
        return 2000

class SmallFan(Fan):
    def power(self):
        return 20
````

It stops with a **TypeError**.

The error is on the line `a = Appliance("Philips")`.
`Appliance` inherits from `ABC` and still has abstract methods (`power` and `describe`).
You cannot create an object of an abstract class, so Python raises a **TypeError**.

## 2. Properties and private attributes

Run this cell first.
It is used by the next four exercises.

````python
class Song:
    def __init__(self, title, seconds):
        self.__title = title
        self.__seconds = seconds

    @property
    def title(self):
        return self.__title

    @property
    def seconds(self):
        return self.__seconds

    @seconds.setter
    def seconds(self, newSeconds):
        if newSeconds > 0:
            self.__seconds = newSeconds

    @property
    def minutes(self):
        return self.__seconds / 60

    def __str__(self):
        return f"{self.__title} ({self.__seconds}s)"
````

### ✍️ Exercise 14.42: Replacement or refinement?

```python
class CashCard:
    def deduct(self, amount):
        if self._balance >= amount:
            self._balance -= amount
            return True
        return False

class MemberCashCard(CashCard):
    def deduct(self, amount):
        if super().deduct(amount):
            self._points += int(amount)
            return True
        return False

    def __str__(self):
        return f"Points: {self._points}"
```

(a) Which type of overriding is `MemberCashCard.deduct`? Explain.

(b) Which type of overriding is `MemberCashCard.__str__`? Explain.

✍️ **My answer:** *(double-click here and type your answer)*

<details>
<summary>👀 Model answer (click to open)</summary>

(a) **Overriding by refinement.**
`deduct` has the same name and parameters as the parent method, and it **calls the parent version** with `super().deduct(amount)`, then adds more work (adding points).

(b) **Overriding by replacement.**
`__str__` overrides the `__str__` of its parent (from `object` or `CashCard`), but it does **not** call `super().__str__()`.
It completely replaces the parent version.

</details>

<details>
<summary>👀 Model answer (click to open)</summary>

Encapsulation means keeping an object's data and the methods that work on that data together inside the class, and **hiding the data** from the outside.

✍️ **My answer:** *(double-click here and type your answer)*

<details>
<summary>👀 Model answer (click to open)</summary>

An abstract class is a general class that is used only as a base for other classes.
No object can be created from it.
In Python it inherits from `ABC`, and it has at least one method marked `@abstractmethod`.
We use it to make sure that every subclass implements certain methods, for example every `Shape` must have an `area()` method, while each subclass calculates it in its own way.

<details>
<summary>👀 Model answer (click to open)</summary>

The bonus rate is the same for **all** cash cards.
