## 1. Why abstract classes?

Imagine someone asks you: "What is the area of a **shape**?"
You cannot answer.
You can only work out the area of a **specific** shape: a circle, a rectangle, a triangle.

So `Shape` is too general to be a real object.
It is a **framework** (a plan) for the more specific classes.

An **abstract class**:

- is a parent class that is **too general** to have objects of its own,
- can have **abstract methods**: methods with a header but no real code (just `pass`),
- forces every child class to write those methods.

You cannot create an object of an abstract class.
You can only create objects of its child classes.

## 2. Writing an abstract class

You need three things.
Learn them by heart:

1. `from abc import ABC, abstractmethod` at the top.
2. The class inherits from `ABC`: `class Shape(ABC):`
3. Put `@abstractmethod` on the line above each abstract method. Its body is just `pass`.

This is the course example.

````python
from abc import ABC, abstractmethod
from math import pi


class Shape(ABC):
    def __init__(self, length):
        self.__length = length

    @property
    def length(self):
        return self.__length

    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__(length)
        self.__width = width

    @property
    def width(self):
        return self.__width

    def area(self):
        return self.length * self.__width


class Circle(Shape):
    def __init__(self, radius):
        super().__init__(radius)

    @property
    def radius(self):
        return self.length

    def area(self):
        return pi * self.radius ** 2


r = Rectangle(2, 10)
c = Circle(2)
print(r.area())
print(f"{c.area():.2f}")
````

Notice:

- `Shape` still has an `__init__` and a property. An abstract class can have normal code too.
- `Rectangle` and `Circle` call `super().__init__(...)` like in notebook 10.
- Each child writes its **own** `area()`. This is how they fill in the plan.

## 3. Abstract methods inside normal methods

An abstract class can have **normal** methods that call its **abstract** methods.
The parent writes the general steps once.
Each child only fills in the small part that is different.

Here a library item works out a fine for a late return.
`calcFine()` is the same for every item.
But the fine **per day** depends on the kind of item, so `getDailyFine()` is abstract.

````python
class LibraryItem(ABC):
    def __init__(self, itemId, title):
        self.__itemId = itemId
        self.__title = title

    @property
    def itemId(self):
        return self.__itemId

    @property
    def title(self):
        return self.__title

    @abstractmethod
    def getLoanDays(self):
        pass

    @abstractmethod
    def getDailyFine(self):
        pass

    def calcFine(self, daysLate):             # a normal method
        if daysLate <= 0:
            return 0
        return daysLate * self.getDailyFine()  # calls the child's version

    def __str__(self):
        return f"[{self.__itemId}] {self.__title}"
````

When `calcFine()` runs for a book, `self` is a book, so `self.getDailyFine()` runs the **book's** version.
You saw the same idea with `describe()` and `getPrice()` in notebook 10.

## 4. Polymorphism

**Polymorphism** is a Greek word that means "**many forms**".

Here is the course example.
One list holds different kinds of shapes.
One line of code, `s.area()`, works for all of them.

````python
shapes = [Circle(2), Rectangle(2, 10), Triangle(10, 5), Square(3)]
for s in shapes:
    print(f"{s.area():.2f}")
````

🎯 **The sentences to learn for the exam:**

> Polymorphism means that the same method call behaves differently depending on the class of the object.
>
> In `for s in shapes: print(s.area())`, the variable `s` is a **polymorphic variable**, because it can refer to a Circle, a Rectangle or a Triangle object.
> The call `s.area()` is **polymorphic**, because which `area()` runs depends on the object that `s` refers to.

Polymorphism works because of inheritance and overriding:
every child class **overrides** the same method in its own way.

"If it walks like a duck and quacks like a duck, it is a duck."

## 📝 Key points

- An **abstract class** is too general to have objects. It is a plan for its child classes.
- Three steps: `from abc import ABC, abstractmethod`, `class X(ABC):`, and `@abstractmethod` above the method (body: `pass`).
- Creating an object of an abstract class raises a **TypeError**.
- A child that does not write **every** abstract method is still abstract (TypeError when you create it).
- A child that **inherits** a real method from its parent is fine.
- An abstract class can have `__init__`, properties and normal methods. Normal methods can call abstract ones.
- **Polymorphism**: the same method call behaves differently depending on the class of the object.
- **Duck typing**: Python only cares that the object has the method, not what class it is.
- **Inheritance** = "is a". **Composition** = "has a".

## ⚠️ Common mistakes

| Mistake | What happens |
| --- | --- |
| Forgetting `(ABC)` after the class name | The class is **not** abstract, so objects can be created |
| Forgetting `@abstractmethod` | That method is not abstract, so children do not have to write it |
| Forgetting to import `ABC` or `abstractmethod` | `NameError` |
| Spelling the child's method differently | The child stays abstract: `TypeError` when you create it |
| Using `self.__weight` (the parent's private attribute) in a child | `AttributeError`. Use the property `self.weight` |
| Answering "polymorphism" without an example | You lose marks. Always add one line of code |
