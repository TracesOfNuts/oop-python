## 1. The "is a" relationship

A student **is a** person.
A lecturer **is a** person.
A tutor **is a** person.

They all have a name, and they can all say hello.
We do not want to write the same name code three times.
So we write it **once**, in a `Person` class, and the other classes **inherit** it.

```text
               Person              <- general (the parent)
          /      |       \
   Lecturer    Tutor    Student    <- specific (the children)
```

New words:

- **Parent class** (also called **superclass** or **base class**): the general class, here `Person`.
- **Child class** (also called **subclass**): the specific class, here `Student`.
- **Inherit**: the child gets all the attributes and methods of the parent, for free.

````python
class Person:
    def __init__(self, name):
        self.__name = name

    @property
    def name(self):
        return self.__name

    def greet(self):
        return f"Hi, I am {self.__name}"


class Student(Person):     # the brackets mean: Student is a Person
    pass                   # nothing new yet


s = Student("Aisha")
print(s.name)
print(s.greet())
````

`Student` has no code inside, only `pass`.
But a `Student` object still has a `name` and a `greet()` method.
They came from `Person`.

`class Student(Person):` is read as "**Student is a Person**".

## 2. Adding new things with `super().__init__()`

A student also has a student ID.
So `Student` needs its **own** `__init__` with one more parameter.

But the name is still stored by `Person`.
Inside the child's `__init__`, we ask the parent to do its part with `super().__init__(...)`.

**`super()` means "my parent class".**

````python
class Student(Person):
    def __init__(self, name, studentId):
        super().__init__(name)          # 1. let Person store the name
        self.__studentId = studentId    # 2. Student stores its own extra data

    @property
    def studentId(self):
        return self.__studentId


s = Student("Aisha", "S1001")
print(s.name, s.studentId)
print(s.greet())
````

How to write a child constructor, every time:

1. The parameters are the parent's ones **plus** the new ones: `(self, name, studentId)`.
2. The **first** line calls `super().__init__(...)` with exactly what the parent's `__init__` needs (without `self`).
3. Then store the new attributes.

💡 A child **adds** to the parent.
`Student` has everything `Person` has, plus `studentId`.

## 3. Overriding a method

Sometimes the child should do something **differently** from the parent.
Then the child writes a method with the **same name and the same parameters**.
This is called **overriding**.
For a child object, Python uses the child's version.

There are two kinds of overriding.
Learn these two sentences by heart:

- **Overriding by replacement**: the child's method does **not** call the parent's version. It replaces it completely.
- **Overriding by refinement**: the child's method **calls the parent's version** with `super().method()`, and then adds more.

````python
class Drink:
    def __init__(self, name, price):
        self.__name = name
        self.__price = price

    @property
    def name(self):
        return self.__name

    def getPrice(self):
        return self.__price

    def describe(self):
        return f"{self.__name} ${self.getPrice():.2f}"


class IcedDrink(Drink):
    def getPrice(self):                    # refinement: starts from the parent's answer
        return super().getPrice() + 0.50


class FreeDrink(Drink):
    def getPrice(self):                    # replacement: ignores the parent completely
        return 0.0


print(Drink("Kopi", 1.80).describe())
print(IcedDrink("Kopi Peng", 1.80).describe())
print(FreeDrink("Water", 1.00).describe())
````

Look closely at `describe()`.
It is written only once, in `Drink`, and it calls `self.getPrice()`.
When `self` is an `IcedDrink`, Python runs the **IcedDrink** version of `getPrice()`.
So the same `describe()` gives a different price for each kind of drink.

💡 Python always starts looking for a method in the object's **own** class.
If it is not there, it looks in the parent.

### Overriding `__str__`

Every class in Python secretly inherits from a built-in class called `object`.
`object` already has a `__str__` method.
That is why printing an object without `__str__` shows something ugly like `<__main__.Drink object at 0x000001>`.

- When **your** class writes `__str__`, it **replaces** the one from `object`. That is overriding by **replacement**.
- When a **child** writes `__str__` that calls `super().__str__()` and adds more, that is overriding by **refinement**.

🎯 Both of these appear in the exam.

## 4. The course example: CashCard and MemberCashCard

Here is the `CashCard` class from the course.
Run it, because the next exercises build on it.

````python
class CashCard:
    def __init__(self, id, amount):
        self.__id = id
        self.__balance = amount

    @property
    def id(self):
        return self.__id

    @property
    def balance(self):
        return self.__balance

    def topUp(self, amount):
        if amount > 0:
            self.__balance += amount

    def deduct(self, amount):
        if self.__balance >= amount:
            self.__balance -= amount
            return True
        return False

    def __str__(self):
        return f"Id: {self.__id} Balance: ${self.__balance:.2f}"


card = CashCard("C1", 20)
print(card.deduct(5))
print(card)
````

A **member cash card** is issued by an organisation, like NTUC.
It **is a** cash card, but it also collects **points**: one point for every whole dollar spent.

```text
         CashCard                      MemberCashCard  (is a CashCard)
   ---------------------          ------------------------------------
   id, balance                    + organisation, points
   topUp(amount)                  (inherits topUp)
   deduct(amount)                 deduct(amount)       <- refinement
   __str__()                      __str__()            <- refinement
                                  + redeemPoints(pts)  <- new method
```

We will build it in three steps.

**Step 2: `deduct` by refinement.**

The parent's `deduct` already checks the balance and takes the money.
We do not want to write that again.
So the child **calls the parent's `deduct`**, and only if it worked, adds the points.

```python
def deduct(self, amount):
    if super().deduct(amount):          # the parent does the money part
        self.__points += int(amount)    # the child adds the points part
        return True
    return False
```

This is exactly the course example of **overriding by refinement**.

**The other way: protected attributes.**

If the parent stores the attribute with **one** underscore, like `self._balance`, it is **protected**.
Python does not rename it, so the child **can** use `self._balance`.

````python
class Account:
    def __init__(self, amount):
        self._balance = amount            # protected: children may use it

    @property
    def balance(self):
        return self._balance


class SavingsAccount(Account):
    def addInterest(self):
        self._balance += self._balance * 0.02    # allowed, because it is protected


acc = SavingsAccount(100)
acc.addInterest()
print(acc.balance)
````

🎯 **In the exam**, the class diagram tells you which one to use.
If it shows `_balance`, use one underscore.
If it shows `__balance`, use two, and in the child use the **property**.

## 6. Class variables in child classes

## 📝 Key points

- Inheritance is the **"is a"** relationship: `class Student(Person):`.
- A child inherits **all** the attributes and methods of its parent.
- In the child's `__init__`, call `super().__init__(...)` **first**, with only what the parent needs.
- **Overriding**: the child writes a method with the same name and parameters.
- **Replacement**: the child's method does not call `super()`. **Refinement**: it calls `super().method()` and adds more.
- A child **cannot** use the parent's `self.__private` attribute. Use the parent's property, or a protected `_attribute`.
- Read class variables with `type(self)._NAME` so a child can have its own value.
- A child object **is also** a parent object: `isinstance(child, Parent)` is `True`.

## ⚠️ Common mistakes

| Mistake | What happens |
| --- | --- |
| Forgetting `super().__init__(...)` | `AttributeError` later, when the parent's data is used |
| Passing the wrong arguments to `super().__init__` | `TypeError` |
| Using `self.__balance` of the parent in the child | `AttributeError` (name mangling) |
| Calling a child-only method on a parent object | `AttributeError` |
| Writing `super.__init__()` without the brackets after `super` | `TypeError` |
| Saying "refinement" without saying **why** | You lose the explanation mark |
