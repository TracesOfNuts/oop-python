## How a notebook works

A notebook is made of **cells**.
There are two kinds:

- **Text cells** (like this one) explain things.
- **Code cells** (with `[ ]:` on the left) hold Python code that you can run.

To run a cell, click on it and press **Shift + Enter**.
The result appears under the cell.
Try it now on the cell below.

````python
print("Hello! I am ready to learn Python.")
````

After a cell runs, a number appears on its left, like `[1]:`.
The number shows the order in which you ran the cells.

⚠️ **Order matters.**
Python remembers what you ran before, not what is above on the page.
If a cell uses something from an earlier cell, run the earlier cell first.

💡 If things get confusing, use the menu **Kernel → Restart Kernel and Run All Cells** (or **Run → Run All Cells**).
This makes Python forget everything and run the notebook again from the top.

Run the two cells below in order.

````python
favouriteDrink = "bubble tea"
````

````python
print("My favourite drink is", favouriteDrink)
````

## Reading an error message

Errors are normal.
Every programmer sees them all day.
An error message is Python **telling you what is wrong**, so learn to read it.

Run the cell below.
It is meant to fail.

````python
price = 3.50
print(prise)
````

How to read the red box:

1. **Look at the last line first.**
   It says `NameError: name 'prise' is not defined`.
   The first word is the **type** of error, and the rest explains it.
2. **Look at the arrow** `---->` or the line number.
   It points to the line that failed.
3. **Ask: what did I mean?**
   Here we typed `prise` instead of `price`.

The most common errors you will meet:

| Error | Usual meaning |
| --- | --- |
| `NameError` | You used a name that does not exist (often a typo, or a cell you forgot to run). |
| `SyntaxError` | Python cannot understand the line (missing `:` or bracket or quote). |
| `IndentationError` | The spaces at the start of a line are wrong. |
| `TypeError` | You used the wrong kind of value, for example adding text to a number. |
| `ValueError` | The kind is right but the value is wrong, for example `int("abc")`. |
| `AttributeError` | The object does not have that attribute or method. |

You will meet all of these again, and the exam asks about them.
