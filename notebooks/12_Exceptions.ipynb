## 1. Three kinds of errors

| Kind | When you see it | Example |
| --- | --- | --- |
| **Syntax error** | Before the program even starts. Python cannot understand the code. | Missing `:` after `if x > 3` |
| **Runtime error** | While the program runs. The code is valid, but something goes wrong. | Dividing by zero, `int("abc")` |
| **Logic error** | Never! The program runs, but the answer is wrong. | Forgetting brackets in an average |

A runtime error is also called an **exception**.
Python creates an **exception object** that describes the problem.
If nobody catches it, the program stops and shows the red error box.

This notebook is about runtime errors, because those are the ones you can **catch** and handle.

## 3. Catching an exception with `try` and `except`

Put the risky code inside `try:`.
If an exception happens, Python **jumps** straight to the `except:` block.
The program then carries on normally instead of crashing.

```python
try:
    # code that might fail
except ValueError:
    # what to do if it fails
```

````python
text = "abc"
try:
    qty = int(text)
    print("You ordered", qty)
except ValueError:
    print("Please type a whole number")
print("The program is still running")
````

Notice that `print("You ordered", qty)` did **not** run.
As soon as `int(text)` failed, Python jumped to `except`.

**Getting the message.**
Write `except ValueError as e:`.
Now `e` is the exception object, and `print(e)` shows its message.

````python
try:
    qty = int("abc")
except ValueError as e:
    print("Error:", e)
````

## 4. The full shape: `try`, `except`, `else`, `finally`

```python
try:
    # risky code
except SomeError as e:
    # runs ONLY if that error happened
else:
    # runs ONLY if NO error happened
finally:
    # ALWAYS runs, error or not
```

`else` and `finally` are optional.

| What happened in `try` | Blocks that run |
| --- | --- |
| No error | `try` (all of it), then `else`, then `finally` |
| An error that is caught | `try` (up to the error), then `except`, then `finally` |

💡 Remember: **`finally` always runs.**
It runs even when there is a `return` inside `try` or `except`.

## 5. Several `except` blocks, and why the order matters

One `try` can have several `except` blocks, one for each kind of error.
Python checks them **from top to bottom** and uses the **first one that matches**.

`except Exception` matches **every** normal error.
So if it comes first, the more specific blocks below it can never run.

**Rule: put the specific exceptions first, and `Exception` last.**

````python
def readPrice(text):
    try:
        price = float(text)
        return 100 / price
    except ValueError:
        return "not a number"
    except ZeroDivisionError:
        return "cannot be zero"
    except Exception:
        return "something else went wrong"

print(readPrice("4"))
print(readPrice("abc"))
print(readPrice("0"))
````

### The exception family tree

Exceptions are classes, and they use **inheritance** (notebook 10).
Here is part of the family tree:

```text
Exception
 +-- ArithmeticError
 |    +-- ZeroDivisionError
 +-- LookupError
 |    +-- IndexError
 |    +-- KeyError
 +-- ValueError
 +-- TypeError
 +-- AttributeError
 +-- NameError
```

## 6. Your own exceptions and `raise`

Python's errors are about Python problems, like dividing by zero.
Your program also has **business rules**, for example:

- A cash card balance cannot go below zero.
- A student cannot join more than 3 CCAs.

When a rule is broken, you **raise your own exception**.

**Step 1: make the exception class.**
It inherits from `Exception`. It needs nothing inside, so the body is just `pass`.

```python

```

**Step 2: raise it with a clear message.**

```python
raise CashCardException("Insufficient balance")
```

## 7. Raising inside a class

In the exam, you raise exceptions inside **methods** of a class.
There are three common places:

1. **In `__init__`**, so a bad object is never created.
2. **In a setter**, so a property cannot be set to a bad value.
3. **In a normal method**, such as `deduct` or `add`.

````python
class CashCardException(Exception):
    pass

class CashCard:
    def __init__(self, cardId, balance):
        if balance < 0:
            raise CashCardException("Balance cannot be negative")
        self.__cardId = cardId
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    def deduct(self, amount):
        if amount > self.__balance:
            raise CashCardException(f"Insufficient balance: ${self.__balance:.2f}")
        self.__balance -= amount

    def __str__(self):
        return f"Id: {self.__cardId} Balance: ${self.__balance:.2f}"

try:
    card = CashCard("C1", 10)
    card.deduct(4)
    print(card)
    card.deduct(50)
    print(card)
except CashCardException as e:
    print("Error:", e)
````

The second `print(card)` never runs, because `deduct(50)` raised the exception first.

💡 Look at the order inside `deduct`: **check first, then change**.
If you subtract first and check afterwards, the balance is already wrong when the error is raised.

## 8. Where to catch: the exception travels up

When a method raises an exception and does **not** catch it, the exception leaves the method.
It goes **up** to the code that called the method.
If that code does not catch it either, it goes up again, until someone catches it or the program crashes.

```text
main()          <- try / except here catches it
  calls addContact()       <- no try here, the exception passes through
    calls Contact()        <- raise happens here
```

🎯 Exam questions often say **"Do not handle the exception in this method."**
That means: **no `try` in that method**.
Just `raise` (or let it pass through), and the caller will handle it.
Writing a `try` there loses marks.

### A menu that survives errors (Lab 4 and JAN 2023 style)

This menu keeps running until the user chooses `3`.
Each option is inside a `try`, so a mistake never ends the program.

````python
class GroupException(Exception):
    pass

class TutorialGroup:
    def __init__(self, name):
        self.__name = name
        self.__students = []

    def addStudent(self, name):
        if name in self.__students:
            raise GroupException(f"{name} already added!")
        self.__students.append(name)

    def removeStudent(self, name):
        if name not in self.__students:
            raise GroupException(f"{name} not found!")
        self.__students.remove(name)

def main():
    group = TutorialGroup("T01")
    while True:
        print("1. Add  2. Remove  3. Quit")
        option = input("Enter option: ")
        if option == "3":
            break
        try:
            if option == "1":
                group.addStudent(input("Name: "))
                print("Added!")
            elif option == "2":
                group.removeStudent(input("Name: "))
                print("Removed!")
            else:
                print("Invalid option")
        except GroupException as e:
            print(e)
    print("Bye")

main()
````

```python
    def __init__(self, studentId, name):
        self.__studentId = studentId
        self.__name = name
        self.__ccas = []

    @property
    def studentId(self):
        return self.__studentId

    def searchCCA(self, ccaName):
        for cca in self.__ccas:
            if cca.name.lower() == ccaName.lower():
                return cca
        return None

    def enrol(self, cca):
        if self.searchCCA(cca.name) is not None:

        self.__ccas.remove(cca)

        for cca in self.__ccas:
```

```python
def main():
    amy = Student("S001", "Amy")
    ccas = [CCA("Chess", 20), CCA("Drama", 35), CCA("CHESS", 20), CCA("Choir", 10), CCA("Soccer", 30)]
    for cca in ccas:
        try:
            amy.enrol(cca)
            print(f"Enrolled in {cca.name}")
        except EnrolmentException as e:
            print(f"Error: {e}")
    for name in ["Golf", "Drama"]:
        try:
            cca = amy.withdraw(name)
            print("Withdrew", cca)
        except EnrolmentException as e:
            print(f"Error: {e}")
    print(amy)

main()
```

## 📝 Key points

- **Syntax** errors stop the program before it starts. **Runtime** errors (exceptions) happen while it runs. **Logic** errors give wrong answers with no message.
- `try:` holds the risky code. `except X as e:` runs only if `X` happened, and `print(e)` shows the message.
- `else` runs only if there was **no** error. `finally` **always** runs, even after a `return`.
- Put specific `except` blocks first and `except Exception` last.
- Your own exception: `class XException(Exception): pass`.
- `raise XException("message")` stops the method at once. It does **not** print anything by itself.
- An exception you do not catch travels **up** to the caller. "Do not handle it here" means no `try` in that method.
- Check first, then change. Follow the order of checks in the question and copy the message text exactly.
- `try` **inside** a loop lets the loop carry on after an error.

## ⚠️ Common mistakes

| Mistake | What goes wrong |
| --- | --- |
| `print("Too many")` instead of `raise ...` | No exception, and the bad change still happens |
| `if len(self.__items) > MAX:` before adding | Lets the list grow one too big. Use `len(...) + 1 > MAX` |
| `except Exception` first | The specific blocks below can never run |
| `class XException:` (no `(Exception)`) | `TypeError` when you raise it |
| A `try` inside the method when told not to handle it | Loses marks, and the caller never hears about the error |
| Typing `3` instead of `Student._MAX_CCA` | Marker looks for the class variable |
| Changing the list, then checking | The data is already wrong when the error is raised |
