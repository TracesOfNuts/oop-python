## 1. What is a list?

A **list** holds many values in one variable, in order.
Think of a shopping list: one piece of paper, many lines.

You write a list with square brackets `[ ]`, and commas between the items.

````python
drinks = ["Milk Tea", "Green Tea", "Coffee"]
prices = [4.80, 3.50, 2.20]
emptyList = []
print(drinks)
print(prices)
print(emptyList)
````

`len()` tells you how many items are in a list.

````python
drinks = ["Milk Tea", "Green Tea", "Coffee"]
print(len(drinks))
````

## 2. Reading one item: the index

Each item has a position number called an **index**.
⚠️ The index **starts at 0**, not 1.

| Index | 0 | 1 | 2 |
| --- | --- | --- | --- |
| Item | `"Milk Tea"` | `"Green Tea"` | `"Coffee"` |

You read an item with `listName[index]`.
A negative index counts from the end: `-1` is the last item.

````python
drinks = ["Milk Tea", "Green Tea", "Coffee"]
print(drinks[0])     # first item
print(drinks[2])     # third item
print(drinks[-1])    # last item
````

If you use an index that does not exist, Python stops with an **IndexError**.

````python
drinks = ["Milk Tea", "Green Tea", "Coffee"]
print(drinks[3])
````

Read the last line: `IndexError: list index out of range`.
The list has 3 items, so the indexes are 0, 1 and 2. There is no index 3.

You can also **change** an item by giving its index a new value.

````python
drinks = ["Milk Tea", "Green Tea", "Coffee"]
drinks[1] = "Lemon Tea"
print(drinks)
````

## 3. Adding and removing items

| What you want | Code | Notes |
| --- | --- | --- |
| Add to the end | `names.append("Eve")` | The most common one |
| Add at a position | `names.insert(0, "Eve")` | Other items move right |
| Remove by value | `names.remove("Ben")` | Removes the first match. `ValueError` if not found |
| Remove by index and get it back | `item = names.pop(1)` | `pop()` with no index removes the last item |
| Remove by index | `del names[1]` | Gives nothing back |
| Is it in the list? | `"Ben" in names` | Gives `True` or `False` |

````python
names = ["Amy", "Ben"]
names.append("Cara")
print(names)
names.insert(0, "Zoe")
print(names)
names.remove("Ben")
print(names)
taken = names.pop(0)
print(taken, names)
print("Amy" in names)
print("Ben" in names)
````

⚠️ `remove()` fails if the value is not in the list.
That is why we usually check with `in` first.

````python
names = ["Amy", "Ben"]
names.remove("Cara")
````

💡 The safe way:

```python
if "Cara" in names:
    names.remove("Cara")
```

🎯 This "check first, then remove" idea is in many exam questions.

## 4. The `for` loop: do something for every item

A `for` loop takes the items of a list **one at a time**.

```python
for drink in drinks:
    print(drink)
```

Read it as: "**for each** `drink` **in** the list `drinks`, do the indented lines".

- `drink` is a new variable. Each time round the loop it holds the next item.
- You choose its name. Pick a name that describes **one** item.
- The line ends with a colon `:`.
- The lines that repeat are **indented** (4 spaces), just like inside a function or an `if`.

````python
drinks = ["Milk Tea", "Green Tea", "Coffee"]
for drink in drinks:
    print("I like", drink)
print("Done")
````

`Done` is printed only once, because that line is **not** indented.
It runs after the loop has finished.

## 5. Recipe 1: the total

To add up all the items, start with a total of `0`, then add each item to it inside the loop.

```python
total = 0
for price in prices:
    total += price
```

````python
prices = [4.8, 3.5, 2.2]
total = 0
for price in prices:
    total += price
    print("added", price, "total is now", total)
print(f"Total: ${total:.2f}")
````

⚠️ `total = 0` must be **before** the loop.
If you put it inside, it goes back to 0 every time.

💡 Python also has `sum(prices)`, but in the exam you usually add up one **field** of each object (like `c.balance`), so you need the loop.
Learn the loop.

## 7. Recipe 3: the largest

To find the largest, remember "the best so far".
Start with the **first** item, then compare each item with the best so far.

````python
marks = [62, 80, 45, 90, 71]
highest = marks[0]
for mark in marks:
    if mark > highest:
        highest = mark
print("Highest:", highest)
````

💡 Why start with `marks[0]` and not `0`?
If all the numbers were negative, starting at `0` would give the wrong answer.

Python also has `max(marks)`.
But in the exam you often need the **item** with the largest value (for example the most expensive product), so learn the loop.

## 8. Recipe 4: search and return the item (or `None`)

This is the most important recipe for the exam.
It is exactly the `searchContact` method from the course:

```python
def searchContact(self, name):
    for c in self.__contacts:
        if c.name == name:
            return c
    return None
```

The idea:

- Look at each item.
- As soon as you find a match, **return it**. The function stops there.
- If the loop finishes without finding anything, return `None` (meaning "not found").

We have not learned classes yet, so here we search a list of strings.

````python
def findName(names, target):
    for name in names:
        if name == target:
            return name
    return None

people = ["Amy", "Ben", "Cara"]
print(findName(people, "Ben"))
print(findName(people, "Zoe"))
````

⚠️ The `return None` must be **after** the loop.
A common mistake is to put an `else: return None` inside the loop.
Then the function gives up after looking at only the first item.

## 15. `while` loops (for menus)

A `for` loop goes through a list.
A `while` loop repeats **as long as a condition is True**.
You only need it for one thing in this course: asking the user again and again.

**Pattern A: keep asking until the user presses Enter.**
When the user just presses Enter, `input()` gives an empty string `""`.

````python
names = []
name = input("Enter a name (press Enter to stop): ")
while name != "":
    names.append(name)
    name = input("Enter a name (press Enter to stop): ")
print(names)
````

⚠️ Do not forget the last line inside the loop, which asks again.
Without it, `name` never changes and the loop runs forever.
(If that happens, press the ■ stop button in the toolbar.)

**Pattern B: a menu with `while True` and `break`.**
`while True` repeats forever, until a `break` stops it.

````python
while True:
    print("1. Say hello")
    print("2. Say bye")
    print("3. Quit")
    option = input("Enter option: ")
    if option == "1":
        print("Hello!")
    elif option == "2":
        print("Bye!")
    elif option == "3":
        break
    else:
        print("Invalid option")
print("Program ended")
````

def classReport(names, marks):
    lines = []
    total = 0
    topIndex = 0
    for i in range(len(names)):
        lines.append(f"{names[i]}: {marks[i]}")
        total += marks[i]
        if marks[i] > marks[topIndex]:
            topIndex = i
    lines.append(f"Average: {total / len(marks):.2f}")
    lines.append(f"Top: {names[topIndex]}")
    return "\n".join(lines)

def searchPhone(names, phones, name):
    for i in range(len(names)):
        if names[i] == name:
            return phones[i]
    return None

def addContact(names, phones, name, phone):
    if name in names:
        return False
    names.append(name)
    phones.append(phone)
    return True

def removeContact(names, phones, name):
    for i in range(len(names)):
        if names[i] == name:
            names.pop(i)
            phones.pop(i)
            return True
    return False

## 📝 Key points

- A list keeps items in order: `names = ["Amy", "Ben"]`. The first index is `0`, the last is `-1`.
- `append` adds to the end. `remove(value)` removes by value. `pop(index)` removes by index and gives the item back.
- Check with `in` before `remove`, or you get a `ValueError`.
- `for item in items:` runs the indented lines once per item.
- The seven recipes:

| Recipe | The key lines |
| --- | --- |
| Total | `total = 0` before the loop, `total += x` inside |
| Count | `count = 0` before, `if ...: count += 1` inside |
| Largest | `best = items[0]`, then `if x > best: best = x` |
| Search | `if match: return item` inside, `return None` **after** the loop |
| Found flag | `found = False`, set `True` and `break` when found |
| Check before adding | `if name in items` and `if len(items) + 1 > MAX` before `append` |
| Filter | `result = []`, `result.append(x)` for the ones you keep |

- `"\n".join(listOfStrings)` makes one string with one item per line.
- `while` repeats while a condition is True. Use it for "until Enter" and menus.

## ⚠️ Common mistakes

| Mistake | What happens |
| --- | --- |
| `names[len(names)]` | `IndexError` (the last index is `len - 1`) |
| `names.remove("Zoe")` when Zoe is not there | `ValueError` |
| `return` inside the loop by accident | The function stops after the first item |
| `else: return None` inside a search loop | Only the first item is checked |
| `total = 0` inside the loop | The total starts again every time |
| Forgetting to ask again inside a `while` loop | The loop never ends |
| `"\n".join([1, 2])` | `TypeError` (join needs strings) |
