## 1. Comparing values

A **comparison** asks a yes or no question.
The answer is always `True` or `False` (a `bool`).

| Symbol | Question | Example | Result |
| --- | --- | --- | --- |
| `==` | is equal to? | `5 == 5` | `True` |
| `!=` | is not equal to? | `5 != 5` | `False` |
| `<` | is less than? | `3 < 5` | `True` |
| `>` | is greater than? | `3 > 5` | `False` |
| `<=` | is less than or equal to? | `5 <= 5` | `True` |
| `>=` | is greater than or equal to? | `4 >= 5` | `False` |

````python
age = 19
print(age >= 18)
print(age == 21)
print("Amy" == "Amy")
````

⚠️ **`=` and `==` are different.**

- `=` **puts** a value into a variable: `age = 19`.
- `==` **asks** whether two values are equal: `age == 19`.

## 3. `else`: do something different when it is false

`else` gives a second path.
Exactly **one** of the two blocks runs.

````python
def ticketType(age):
    if age < 12:
        return "Child"
    else:
        return "Adult"

print(ticketType(8))
print(ticketType(30))
````

💡 Notice the indentation inside a function.
The `if` is indented once (it is inside the function).
The `return` lines are indented twice (they are inside the `if`).

## 4. `elif`: more than two choices

`elif` means "else, if".
Python checks the conditions **from top to bottom**.
It runs the block of the **first** condition that is true, and skips all the rest.
`else` at the end catches everything that did not match.

````python
def drinkSize(ml):
    if ml >= 700:
        return "Large"
    elif ml >= 500:
        return "Medium"
    else:
        return "Small"

print(drinkSize(750))
print(drinkSize(500))
print(drinkSize(300))
````

## 6. Tools for checking text

Strings have built-in **methods**.
A method is a function that belongs to a value.
You call it with a dot: `text.lower()`.

| Tool | What it does | Example | Result |
| --- | --- | --- | --- |
| `text.lower()` | all small letters | `"Tea".lower()` | `"tea"` |
| `text.upper()` | all capital letters | `"Tea".upper()` | `"TEA"` |
| `text.strip()` | removes spaces at both ends | `"  hi  ".strip()` | `"hi"` |
| `len(text)` | number of characters | `len("hello")` | `5` |
| `part in text` | is `part` inside `text`? | `"ea" in "Tea"` | `True` |
| `text.startswith(part)` | does `text` begin with `part`? | `"91234567".startswith("9")` | `True` |
| `text.isdigit()` | is every character a digit 0 to 9? | `"123".isdigit()` | `True` |

⚠️ These methods do **not** change the original text.
They give back a new value, so you must use or store the result.

## 7. Comparing text without caring about capitals

🎯 Exam questions often say "search by name, **case insensitive**".
That means `"amy"`, `"Amy"` and `"AMY"` must all count as the same name.
The trick: make **both** sides lowercase before you compare.

````python
storedName = "Amy"
typedName = "AMY"
print(storedName == typedName)                  # False
print(storedName.lower() == typedName.lower())  # True
````

## 📝 Key points

- Comparisons (`==`, `!=`, `<`, `>`, `<=`, `>=`) give `True` or `False`.
- `=` stores a value. `==` asks a question.
- `if condition:` then an indented block. Use `elif` and `else` for other cases.
- In an `if` / `elif` chain, only the **first** true branch runs, so order matters.
- `and` needs both, `or` needs one, `not` flips. Use brackets when mixing `and` and `or`.
- For case insensitive checks, compare `a.lower() == b.lower()`.
- Clean typed text with `.strip()` before checking it.
- "Results in more than the maximum" means `count + 1 > MAX`.

## ⚠️ Common mistakes

| Mistake | What happens |
| --- | --- |
| `if x = 5:` | `SyntaxError` (use `==`) |
| Missing `:` after `if`, `elif` or `else` | `SyntaxError` |
| Wrong indentation under `if` | `IndentationError`, or a line runs when it should not |
| `day == "Sat" or "Sun"` | Always counts as true. Write `day == "Sat" or day == "Sun"` |
| `"Amy" == "amy"` | `False`. Use `.lower()` on both sides |
| `if count > MAX:` when the rule is about adding one more | Allows one item too many. Use `count + 1 > MAX` |
| Most general check first in an `elif` chain | The specific case is never reached |
